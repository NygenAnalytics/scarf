```python
from pathlib import Path

import scarf
import scarf.plotting as splt

DATASET = "bastidas-ponce_4K_pancreas-d15_rnaseq"
repo_root = Path(scarf.__file__).resolve().parents[1]
zarr_path = (
    repo_root
    / "docs"
    / "source"
    / "vignettes"
    / "scarf_datasets"
    / DATASET
    / "data.zarr"
)
if not zarr_path.exists():
    scarf.fetch_dataset(
        dataset_name=DATASET,
        save_path="scarf_datasets",
        as_zarr=True,
    )
    zarr_path = Path("scarf_datasets") / DATASET / "data.zarr"

ds = scarf.DataStore(str(zarr_path), nthreads=4, default_assay="RNA")
```

```python
splt.embedding(ds, layout_key="RNA_UMAP", color_by="clusters").figure
```

```python
splt.embedding(
    ds,
    layout_key="RNA_UMAP",
    color_by=["Gcg", "Ins2", "Sst"],
    normalization=splt.NormalizationSpec(transform="log1p"),
    sort_values=True,
).figure
```

```python
splt.embedding(
    ds,
    layout_key="RNA_UMAP",
    color_by="Gcg",
    normalization=splt.NormalizationSpec(transform="log1p"),
    color_scale=splt.ColorScale(cmap="viridis", quantiles=(0.0, 0.99)),
    sort_values=True,
).figure
```

```python
splt.embedding_raster(
    ds,
    layout_key="RNA_UMAP",
    color_by="RNA_nCounts",
    pixels=400,
).figure
```

```python
splt.embedding(
    ds,
    layout_key="RNA_UMAP",
    color_by="clusters",
    legend_loc="on_data",
).figure
```

```python
splt.embedding(
    ds,
    layout_key="RNA_UMAP",
    color_by="clusters",
    legend_loc="on_data",
    frame="none",
    theme="paper",
).figure
```

```python
n = len(ds.cells.active_index("I"))
ds.cells.insert(
    "demo_sample",
    [f"s{i % 8}" for i in range(n)],
    overwrite=True,
)

splt.dotplot(
    ds,
    features={"endocrine": ["Gcg", "Ins2", "Sst"]},
    group_by="clusters",
    sample_by="demo_sample",
).figure
```

```python
splt.matrixplot(
    ds,
    features=["Gcg", "Ins2", "Sst"],
    group_by="clusters",
    value="mean",
).figure
```

```python
ds.cells.insert(
    "demo_subject",
    [f"d{i % 4}" for i in range(n)],
    overwrite=True,
)
ds.cells.insert(
    "demo_condition",
    ["before" if i % 8 < 4 else "after" for i in range(n)],
    overwrite=True,
)

splt.composition(
    ds,
    category_by="clusters",
    study_design=splt.StudyDesign(
        sample_by="demo_sample",
        subject_by="demo_subject",
        condition_by="demo_condition",
    ),
    kind="per_sample",
).figure
```

```python
splt.distribution(
    ds,
    keys=["RNA_nCounts", "RNA_nFeatures"],
    group_by="clusters",
    kind="violin",
    max_points=2000,
    seed=0,
).figure
```

```python
from pathlib import Path

out = Path("scarf_datasets") / "plotting_showcase_embedding.png"
out.parent.mkdir(parents=True, exist_ok=True)
result = splt.embedding(
    ds,
    layout_key="RNA_UMAP",
    color_by="clusters",
    show=False,
)
result.save(out, dpi=200)
assert out.exists()
result.close()
```
