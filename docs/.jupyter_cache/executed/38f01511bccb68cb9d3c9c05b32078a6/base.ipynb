```python
import scarf

dataset = scarf.cytebase.connect("scarf_docs").download_dataset(
    "tenx_5K_pbmc_rnaseq",
    destination="scarf_datasets",
    zarr=True,
)
ds = scarf.DataStore(f"{dataset}/data.zarr", nthreads=4)
run = ds.pipeline.open(label="docs_default")
```

```python
ds.plots.embedding(
    run=run,
    color_by="clusters",
    legend_loc="on_data",
)
```
